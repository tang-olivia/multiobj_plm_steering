```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def compute_mean_tm(csv_path):
    df = pd.read_csv(csv_path)
    return df['Tm'].mean()

no_steering_path = '/home/ubuntu/68610_project/evaluation_res/no_steering.csv'
thermo_all_layers_path = '/home/ubuntu/68610_project/evaluation_res/thermo_steering_all_layers_a05.csv'

no_steering_easy_tm = compute_mean_tm(no_steering_path)
thermo_all_layers_easy_tm = compute_mean_tm(thermo_all_layers_path)

print(f"No steering easy: {no_steering_easy_tm}")
print(f"Thermo all layers easy: {thermo_all_layers_easy_tm}")
```

```python
def plot_distribution(csv_path, title, png_path):
    df = pd.read_csv(csv_path)
    tm_l = df['Tm'].tolist()
    plt.hist(tm_l, bins=20, edgecolor='black')
    plt.xlabel('Tm (°C)')
    plt.ylabel('Frequency')
    plt.title(title)
    plt.savefig(png_path)
```

```python
plot_distribution(no_steering_path, 'Tm distribution (no steering)', '/home/ubuntu/68610_project/figures/no_steering_tm_dist.png')
```

```python
plot_distribution(thermo_all_layers_path, 'Tm distribution (thermo-only, alpha=0.5)', '/home/ubuntu/68610_project/figures/thermo_only_a05_tm_dist.png')
```
